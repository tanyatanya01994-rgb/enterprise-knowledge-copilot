# RAG Evaluation

## Enterprise Knowledge Copilot

This notebook documents the evaluation methodology used for the Enterprise Knowledge Copilot.

The evaluation dataset contains different types of questions to test retrieval quality, grounded generation, follow-up handling, multi-document reasoning, and safe handling of unsupported questions.

### Evaluation Categories

- Supported questions
- Unsupported questions
- Follow-up questions
- Multi-document questions

### Evaluation Dimensions

- Retrieval relevance
- Context relevance
- Answer correctness
- Groundedness
- Completeness
- Hallucination prevention
- Safe abstention

The production evaluation runner is located in `evaluation/run_evaluation.py`.

## Safety-Aware Evaluation

A key requirement of the system is to avoid fabricating information.

When the retrieved evidence is insufficient, the system should safely abstain instead of generating an unsupported answer.

The evaluation pipeline also distinguishes external verifier or API failures from genuine unsupported questions. This prevents temporary model availability problems from being incorrectly reported as retrieval failures.

### Evaluation Dataset

The evaluation set includes questions covering:

- Annual leave
- Work from home
- Employee benefits
- Resignation and exit
- Code of conduct
- Attendance
- Compensation
- Follow-up questions
- Multi-document comparisons
- Unsupported questions

```python
import json
from pathlib import Path

evaluation_file = Path("../evaluation/questions.json")

if evaluation_file.exists():
    questions = json.loads(
        evaluation_file.read_text(encoding="utf-8")
    )

    print(f"Evaluation questions: {len(questions)}")

    categories = {}

    for question in questions:
        category = question.get("type", "unknown")
        categories[category] = categories.get(category, 0) + 1

    print("\nQuestion categories:")

    for category, count in sorted(categories.items()):
        print(f"{category}: {count}")
else:
    print("Evaluation dataset not found.")
```
