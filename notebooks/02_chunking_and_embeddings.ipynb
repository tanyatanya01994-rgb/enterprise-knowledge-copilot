# Chunking and Embeddings

## Enterprise Knowledge Copilot

This notebook documents the chunking and embedding stage of the RAG pipeline.

### Chunking Configuration

The project uses structure-aware, page-aware chunks with:

- Chunk size: 120
- Chunk overlap: 30
- Page number metadata
- Document name
- Document type
- Section
- Year

The production implementation is located in `src/chunker.py`.

## Embedding Model

The system uses the Sentence Transformers model:

`all-MiniLM-L6-v2`

The generated dense embeddings are stored in a local Qdrant vector database.

The vector database uses cosine similarity for semantic retrieval.

The production implementation is located in `src/embedder.py`.


```python
import json
from pathlib import Path

chunks_file = Path("../data/processed/chunks.jsonl")

if chunks_file.exists():
    chunks = [
        json.loads(line)
        for line in chunks_file.read_text(encoding="utf-8").splitlines()
        if line.strip()
    ]

    print(f"Processed chunks: {len(chunks)}")

    if chunks:
        print("\nExample chunk metadata:")

        for key, value in chunks[0].items():
            print(f"{key}: {value}")
else:
    print("Processed chunks file not found.")
```
